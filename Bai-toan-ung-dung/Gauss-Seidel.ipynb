```python
import numpy as np

def lap_he(m, in_ra=True):
    n = m * m
    A = np.zeros((n, n))
    b = np.zeros(n)
    for i in range(m):              # i = 0 là hàng sát cạnh trên
        for j in range(m):
            k = i * m + j
            A[k, k] = 4
            if i > 0:     A[k, k - m] = -1
            else:         b[k] += 100        # biên trên 100°C chuyển sang vế phải
            if i < m - 1: A[k, k + m] = -1
            if j > 0:     A[k, k - 1] = -1
            if j < m - 1: A[k, k + 1] = -1

    if in_ra:
        print(f"Hệ {n} phương trình ({n} ẩn), m = {m}")
        print("Ma trận A:")
        print(A.astype(int))
        print("Vế phải b:")
        print(b.astype(int))
    return A, b


def Gauss_Seidel(A, b, max_blap, esl):
    n = len(b)
    m = int(round(np.sqrt(n)))      
    x = np.zeros(n)

    q_values = []
    for i in range(n):
        sigma = np.sum(np.abs(A[i, :])) - np.abs(A[i, i])
        q = round(sigma / np.abs(A[i, i]), 6)
        q_values.append(sigma / np.abs(A[i, i]))
        print(f"q{i + 1} = {q:.6f}")
    t = max(q_values)
    print(f"Giá trị q lớn nhất: {t:.6f}")
    htu = t < 1
    print("Điều kiện q < 1:", htu)
    if not htu:
        print("Chéo trội không ngặt (q = 1), vẫn chạy tiếp.")
    print()

    for k in range(max_blap):
        x_old = np.copy(x)

        for i in range(n):
            sum1 = np.dot(A[i, :i], x[:i])           
            sum2 = np.dot(A[i, i + 1:], x_old[i + 1:])  
            x[i] = (b[i] - sum1 - sum2) / A[i, i]

        saiso = np.max(np.abs(x - x_old))

        print(f"KQ lần lặp thứ {k + 1}:")
        for hang in x.reshape(m, m):
            print("  " + "  ".join(f"{v:9.6f}" for v in hang))
        print(f" Sai số: {saiso:.6f}\n")

        if saiso < esl:
            return x, k + 1, saiso

    return x, max_blap, saiso

# Giải hệ phương trình với m = 5
A, b = lap_he(5, in_ra=False)
kq = Gauss_Seidel(A, b, max_blap=5, esl=1e-5)
```
