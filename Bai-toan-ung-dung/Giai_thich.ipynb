# Giải thích vì sao các phương pháp lặp vẫn hội tụ khi ma trận chéo trội không ngặt
Trong bài toán phân bố nhiệt độ trên tấm phẳng, hệ phương trình tuyến tính được biểu diễn dưới dạng:

$$
Ax=b
$$

Ma trận hệ số $A$ có các phần tử trên đường chéo chính bằng $4$, còn các phần tử ngoài đường chéo khác $0$ đều bằng $-1$.

## 1. Kiểm tra tính chéo trội của ma trận

Theo điều kiện chéo trội, ta có:

$$
|a_{ii}|\geq\sum_{j\neq i}|a_{ij}|
$$

Đối với các điểm nằm sát biên, số điểm lân cận thuộc lưới nhỏ hơn $4$, nên:

$$
|a_{ii}|>\sum_{j\neq i}|a_{ij}|
$$

Do đó, các hàng tương ứng thỏa mãn điều kiện chéo trội ngặt.

Tuy nhiên, tại những điểm có đủ $4$ điểm lân cận bên trong lưới:

$$
|a_{ii}|=4=\sum_{j\neq i}|a_{ij}|
$$

Vì vậy, ma trận $A$ chỉ chéo trội không ngặt trên toàn bộ ma trận.

Đặt:

$$
q=\max_i\frac{\sum_{j\neq i}|a_{ij}|}{|a_{ii}|}
$$

$$
q'=\max_j\frac{\sum_{i\neq j}|a_{ij}|}{|a_{jj}|}
$$

Với $m=3,5,10,20$, ta có:

$$
\boxed{q=q'=1}
$$

Do đó, không thể sử dụng điều kiện $q<1$ hoặc $q'<1$ để kết luận sự hội tụ.

## 2. Kiểm tra tính đối xứng

Trong bài toán phân bố nhiệt độ, nếu điểm thứ $i$ là điểm lân cận của điểm thứ $j$ thì điểm thứ $j$ cũng là điểm lân cận của điểm thứ $i$.

Vì vậy:

$$
a_{ij}=a_{ji}
$$

Suy ra:

$$
\boxed{A=A^T}
$$

Như vậy, ma trận $A$ đối xứng.

## 3. Giải thích sự hội tụ của phương pháp Jacobi

Mặc dù ma trận $A$ không chéo trội ngặt trên tất cả các hàng, phương pháp Jacobi vẫn hội tụ.

Nguyên nhân là:

- Ma trận $A$ chéo trội không ngặt.
- Tồn tại các hàng chéo trội ngặt, tương ứng với những điểm nằm sát biên.
- Các điểm bên trong lưới có sự liên kết với nhau thông qua các điểm lân cận.
- Từ một điểm bất kỳ trong lưới, luôn tồn tại đường đi qua các điểm lân cận đến một điểm nằm sát biên.

Nhờ cấu trúc liên kết này, ảnh hưởng của các điểm có điều kiện chéo trội ngặt được truyền đến toàn bộ các điểm bên trong lưới qua các bước lặp.

Theo định lý hội tụ của phương pháp Jacobi đối với ma trận chéo trội yếu có cấu trúc liên thông và có ít nhất một hàng chéo trội ngặt, phương pháp Jacobi hội tụ.

Do đó:

$$
\boxed{\text{Phương pháp Jacobi hội tụ}}
$$

## 4. Giải thích sự hội tụ của phương pháp Gauss–Seidel

Ma trận $A$ có các tính chất:

- Đối xứng.
- Các phần tử trên đường chéo chính đều dương.
- Chéo trội không ngặt.
- Có các hàng chéo trội ngặt.
- Các điểm trong lưới liên kết với nhau.

Với cấu trúc của bài toán phân bố nhiệt độ, ma trận $A$ là ma trận đối xứng xác định dương.

Tức là:

$$
\boxed{x^TAx>0,\quad \forall x\neq0}
$$

Theo định lý hội tụ của phương pháp Gauss–Seidel, nếu ma trận hệ số đối xứng xác định dương thì phương pháp hội tụ.

Vì vậy:

$$
\boxed{\text{Phương pháp Gauss–Seidel hội tụ}}
$$

## 5. Giải thích sự hội tụ của phương pháp Conjugate Gradient

Theo lý thuyết, phương pháp Conjugate Gradient hội tụ khi ma trận hệ số $A$ đối xứng xác định dương.

Từ các kết quả đã chứng minh ở trên, ta có:

Thứ nhất, ma trận $A$ đối xứng:

$$
A=A^T
$$

Thứ hai, ma trận $A$ xác định dương:

$$
x^TAx>0,\quad\forall x\neq0
$$

Suy ra:

$$
\boxed{A\text{ là ma trận đối xứng xác định dương}}
$$

Như vậy, ma trận $A$ thỏa mãn đầy đủ điều kiện hội tụ của phương pháp Conjugate Gradient.

Theo định lý hội tụ của Conjugate Gradient, đối với ma trận đối xứng xác định dương, phương pháp sẽ tìm được nghiệm chính xác của hệ phương trình sau không quá $n$ bước lặp trong điều kiện số học chính xác.

Trong bài toán này:

$$
n=m^2
$$

Do đó:

$$
\boxed{\text{Phương pháp Conjugate Gradient hội tụ}}
$$

Sự hội tụ này không phụ thuộc vào việc ma trận $A$ có chéo trội ngặt hay không, mà được đảm bảo bởi tính đối xứng xác định dương.
