```python
import numpy as np

def lap_he(m):
    n = m * m
    A = np.zeros((n, n))
    b = np.zeros(n)
    for i in range(m):              # i = 0 là hàng sát cạnh trên
        for j in range(m):
            k = i * m + j
            A[k, k] = 4
            if i > 0:     A[k, k - m] = -1
            else:         b[k] += 100        # biên trên 100°C chuyển sang vế phải
            if i < m - 1: A[k, k + m] = -1
            if j > 0:     A[k, k - 1] = -1
            if j < m - 1: A[k, k + 1] = -1

    return A, b


def Jacobi(A, b, max_lap, esl, in_chi_tiet=True):
    n = len(b)
    m = int(round(np.sqrt(n)))      # kích thước lưới m x m
    x = np.zeros(n)

    # Keep full precision for q; round only for display.
    q_values = []
    for i in range(n):
        sigma = np.sum(np.abs(A[i, :])) - np.abs(A[i, i])
        q = sigma / np.abs(A[i, i])
        q_values.append(q)
        if in_chi_tiet:
            print(f"q{i + 1} = {q:.6f}")

    t = max(q_values)
    print(f"Giá trị q lớn nhất: {t:.6f}")

    htu = t < 1
    print("Điều kiện q < 1:", htu)
    if not htu:
        print("Chéo trội không ngặt (q = 1), vẫn chạy tiếp.")

    print()

    so_lan = 0
    saiso = np.inf
    hoi_tu = False
    for k in range(max_lap):
        x_old = np.copy(x)

        for i in range(n):
            sum1 = np.dot(A[i, :i], x_old[:i])
            sum2 = np.dot(A[i, i + 1:], x_old[i + 1:])
            x[i] = (b[i] - sum1 - sum2) / A[i, i]

        # Sai số: ||x^(k) - x^(k-1)||_inf
        saiso = np.max(np.abs(x - x_old))
        so_lan = k + 1

        # In kết quả theo dạng lưới m x m
        if in_chi_tiet:
            print(f"KQ lần lặp thứ {k + 1}:")
            for hang in x.reshape(m, m):
                print("  " + "  ".join(f"{v:9.6f}" for v in hang))
            print(f" Sai số: {saiso:.3e}\n")

        # Điều kiện dừng: sai số nhỏ hơn ngưỡng esl
        if saiso < esl:
            hoi_tu = True
            break

    if hoi_tu:
        print(f"Hội tụ sau {so_lan} lần lặp (ngưỡng {esl:.0e}), sai số cuối = {saiso:.3e}")
    else:
        print(f"Chưa hội tụ sau {max_lap} lần lặp, sai số = {saiso:.3e}")

    return x, so_lan
```

```python
# Giải hệ phương trình với m = 5, chạy đến khi hội tụ
A, b = lap_he(5)
kq, so_lan = Jacobi(A, b, max_lap=20000, esl=1e-6)
```
