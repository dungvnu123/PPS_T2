```python
import numpy as np

def Jacobi(A, b, max_blap, esl):
    n = len(b)
    x = np.zeros(n)  # Khởi tạo vector x với giá trị 0

    # Kiểm tra điều kiện hội tụ và tìm giá trị q lớn nhất
    q_values = []
    for i in range(n):
        sigma = np.sum(np.abs(A[i, :])) - np.abs(A[i, i])
        q = sigma / np.abs(A[i, i]) if np.abs(A[i, i]) != 0 else 0
        q_values.append(q)
        print(f"q{i + 1} = {q}")
    
    t = max(q_values)
    print("Giá trị q lớn nhất:", t)

    # Kiểm tra điều kiện hội tụ cho tất cả hàng
    htu = t < 1
    print("Điều kiện hội tụ:", htu)
    if not htu:
        print("Hệ phương trình không hội tụ. Dừng thuật toán.")
        return None
        
    print('\n')
        
    for k in range(max_blap):
        x_old = np.copy(x)  # Lưu giá trị cũ để tính toán

        for i in range(n):
            sum1 = np.dot(A[i, :i], x_old[:i])              # Tính tổng phần trước
            sum2 = np.dot(A[i, i + 1:], x_old[i + 1:])  # Tính tổng phần sau
            x[i] = (b[i] - sum1 - sum2) / A[i, i]       # Cập nhật giá trị x[i]

        # Tính sai số
        saiso = np.max(np.abs(x - x_old))
        ketqua = np.array2string(x, formatter={'float_kind': lambda value: f'{value:.6f}'})
        print(f"KQ lần lặp thứ {k + 1}: {ketqua} \n   Sai số: {saiso:.6f}")

        # Kiểm tra nếu sai số nhỏ hơn giá trị tối thiểu
        if saiso < esl:
            print(f"Hội tụ đạt được sau {k + 1} lần lặp.")
            return x

    return x

# Ví dụ
A = np.array([[5, -1, 2],
              [1, 4, -1],
              [1, 1, 4]], dtype=float)
b = np.array([8, -4, 4], dtype=float)

kq = Jacobi(A, b, max_blap=10, esl=1e-5)
print('Nghiệm gần đúng:', np.array2string(kq, formatter={'float_kind': lambda value: f'{value:.6f}'}))
```
