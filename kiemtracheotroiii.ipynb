```python
import numpy as np
def analyze_matrix(A) -> dict:#khai báo
    #ma trận A
    # Chuyển dữ liệu đầu vào thành ma trận số thực bằng cách tự duyệt
    M = []
    for row in A:
        new_row = []
        for value in row:
            new_row.append(float(value))
        M.append(new_row)

    #loại trường hợp ma trận không vuông
    if len(M) == 0:
        raise ValueError("A phải là ma trận vuông")
    n = len(M)
    for row in M:
        if len(row) != n:
            raise ValueError("A phải là ma trận vuông")

    #kiểm tra trường hợp phần tử đường chéo =0
    diag = []
    for i in range(n):
        value = M[i][i]
        if value < 0:
            value = -value
        diag.append(value)
    for value in diag:
        if value < 1e-12:
            raise ValueError("phần tử đường chéo bằng 0, không tính được q và q'")

    #tính tổng ngoài đường chéo bruh
    abs_A = []
    for i in range(n):
        row = []
        for j in range(n):
            value = M[i][j]
            if value < 0:
                value = -value
            row.append(value)
        abs_A.append(row)

    off_row = []
    off_col = []
    for i in range(n):
        total = 0.0
        for j in range(n):
            if i != j:
                total += abs_A[i][j]
        off_row.append(total)

    for j in range(n):
        total = 0.0
        for i in range(n):
            if i != j:
                total += abs_A[i][j]
        off_col.append(total)

    #kiểm tra chéo trội theo hàng go bruh bruh
    row_dominant = True
    row_strict = True
    for i in range(n):
        if diag[i] < off_row[i] - 1e-12:
            row_dominant = False
        if diag[i] <= off_row[i] + 1e-12:
            row_strict = False

    #kiểm tra chéo trội theo cột go bruh bruh
    col_dominant = True
    col_strict = True
    for i in range(n):
        if diag[i] < off_col[i] - 1e-12:
            col_dominant = False
        if diag[i] <= off_col[i] + 1e-12:
            col_strict = False

    #tính q
    q = 0.0
    for i in range(n):
        value = off_row[i] / diag[i]
        if value > q:
            q = value

    #tính q guass seidel q_gs
    
    #tổng trái theo công thức tính tay
    mauso = []
    for i in range(n):
        tong_trai = 0.0
        for j in range(i):
            tong_trai += abs_A[i][j]
        mauso.append(diag[i] - tong_trai)

    q_gs = 0.0
    mau_so_khong_hop_le = False
    for value in mauso:
        if value <= 1e-12:
            mau_so_khong_hop_le = True
    if mau_so_khong_hop_le:
        q_gs = float("inf")
    else:
        
        #tổng phải theo công thức tính tay
        for i in range(n):
            tong_phai = 0.0
            for j in range(i + 1, n):
                tong_phai += abs_A[i][j]
            value = tong_phai / mauso[i]
            if value > q_gs:
                q_gs = value

    ## chuẩn Bvc
    B = []
    for i in range(n):
        row = []
        for j in range(n):
            if j != i:
                row.append(-M[i][j] / M[i][i])
            else:
                row.append(0.0)
        B.append(row)

    chuanvcB = 0.0
    for i in range(n):
        tong = 0.0
        for j in range(n):
            value = B[i][j]
            if value < 0:
                value = -value
            tong += value
        if tong > chuanvcB:
            chuanvcB = tong

    #kiểm tra đối xứng
    dx = True
    for i in range(n):
        for j in range(n):
            if M[i][j] - M[j][i] > 1e-12 or M[j][i] - M[i][j] > 1e-12:
                dx = False

    #bán kính phổ
    eigenvalues = np.linalg.eigvals(B)
    bkpho = 0.0
    for value in eigenvalues:
        magnitude = abs(value)
        if magnitude > bkpho:
            bkpho = magnitude
    # làm tròn 6 chữ số
    bkpho = int(bkpho * 1000000 + 0.5) / 1000000

    return {
        "row_dominant": row_dominant,
        "row_strict": row_strict,
        "col_dominant": col_dominant,
        "col_strict": col_strict,
        "q": lam_tron_6(q),
        "q_gaussseidel": lam_tron_6(q_gs) if q_gs != float("inf") else float("inf"),
        "chuanB": lam_tron_6(chuanvcB),
        "doixungg": dx,
        "Bankinhpho": bkpho,  ## đã làm tròn 6 chữ số
        "matranB": B,
    }
```

```python

# hàm kiểm tra 2 số có GẦN BẰNG nhau hay không, để đảm bảo sai số
def gan_bang(a, b, sai_so=1e-6):
    hieu = a - b
    if hieu < 0:
        hieu = -hieu
    return hieu <= sai_so

# hàm làm tròn đến 6 chữ số sau dấu phẩy 
def lam_tron_6(x):
    if x >= 0:
        return int(x * 1000000 + 0.5) / 1000000 
    # nhân với 1 10^6 vì muốn làm tròn đến 6 chữ số thì nhân với 10^6
    # sau đó làm tròn về số nguyên và lại chia cho 10^6 để được số thập phân có 6 chữ số sau dẩu phẩy
    #vd: 2,1234567 
    # nhân với 10^6 là 2123456,7 
    # làm tròn về số nguyên: 2123457
    # chia 10^6: 2,123457
    return -int((-x) * 1000000 + 0.5) / 1000000
# cộng với 0,5 vì: trong python, int chuyển số thực về số nguyên sẽ bỏ phần thập phân, k làm tròn
#vd: 3,2-->3 và 3.8-->3 
# vì vậy để làm tròn ta cộng thêm 0,5 . Giống như làm tròn đến hàng phần 10

# kiểm tra 2 ma trận có gần bằng nhau hay k
#mục đích: ma trận python tính được và ma trận tính tay có cho kết quả gần giống nhau hay k. để kiểm thử test #1
def assert_ma_tran_gan_bang(A, B, sai_so=1e-6):
    assert len(A) == len(B) # số hàng ma trận A có bằng số hàng ma trận B
    for i in range(len(A)):
        assert len(A[i]) == len(B[i]) # số phần tử ở hàng i  có bằng nhau k
        #A= 1 2 3
        #B= 1 2    -> lỗi 
        for j in range(len(A[i])):
            assert gan_bang(A[i][j], B[i][j], sai_so) # phần từ ij của A gần bằng phần tử ij của B


def assert_loi(func, loai_loi): # khai báo, func: hàm muốn kiểm tra có lỗi hay k, loai_loi: loại lỗi mong muốn

    da_bao_loi = False # tạo biến da_bao_loi và gán giá trị ban đầu là false, đổi thành true nếu có lỗi
    try:
        func()
    except loai_loi:
        da_bao_loi = True # đúng lỗi mong đợi thì true
    assert da_bao_loi


# assert.... thỏa mãn nếu điều kiện đúng, sai thì test thất bại
# ma trận B[i][j] = −A[i][j] / A[i][i] khi j ≠ i, và B[i][i] = 0


# 1 ví dụ chuẩn (tính tay được)
def test_vi_du_chuan():
    A = [[4, 1, 1],
         [1, 5, 2],
         [0, 1, 3]]
    r = analyze_matrix(A)  # hàng 1: 4>2 | hàng 2: 5>3 | hàng 3: 3>1 -> chéo trội nghiêm ngặt
                           # cột 3: 3 = 3 nên trội nhưng không nghiêm ngặt
    assert r["row_dominant"]
    assert r["row_strict"]          # hàng 1: 4>2 | hàng 2: 5>3 | hàng 3: 3>1 -> chéo trội nghiêm ngặt
    assert r["col_dominant"]
    assert not r["col_strict"]
    assert gan_bang(r["q"], 0.6)
    assert gan_bang(r["q_gaussseidel"], 0.5)
    assert gan_bang(r["chuanB"], 0.6)
    assert not r["doixungg"]

    expected_B = [[0, -0.25, -0.25],  # hàng 0
                  [-0.2, 0, -0.4],    # hàng 1
                  [0, -1/3, 0]]       # hàng 2
    assert_ma_tran_gan_bang(r["matranB"], expected_B)


# 2 ma trận đường chéo

# 2  0
# 0  5
def test_ma_tran_duong_cheo():
    r = analyze_matrix([[2, 0], [0, 5]])
    assert r["q"] == 0.0               # q=0 (tổng phần tử ngoài đường chéo / đường chéo)
    assert r["q_gaussseidel"] == 0.0   # q_gs=0
    assert r["Bankinhpho"] == 0.0      # bk phổ = 0
    assert r["doixungg"]               # ma trận đường chéo đối xứng
    assert r["row_strict"]
    assert r["col_strict"]


# 3 ma trận đối xứng 2x2
def test_doi_xung_2x2():

    # A=   2  1
    #      1  2

    r = analyze_matrix([[2, 1], [1, 2]])
    # B= 0     -0,5
    #   -0,5    0
    assert r["doixungg"]
    assert gan_bang(r["q"], 0.5)
    assert gan_bang(r["q_gaussseidel"], 0.5)  # tổng phải / ( đường chéo - tổng trái )
    assert gan_bang(r["Bankinhpho"], 0.5)     # giá trị riêng của B: 0,5
    # Giá trị riêng lamda tạm kí hiệu là: h
    # giải det(B-hI)=0    I là ma trận đơn vị  1 0
    #                                          0 1
    # có hI=   h      0
    #          0      h
    # B-hI =   -h      -0,5
    #          -0,5    -h
    # tính det(B-hI)= h^2 - 0,25 =0 => h= +- 0,5. Vậy giá trị riêng là ±0,5


# 4 không chéo trội
def test_khong_cheo_troi():
    r = analyze_matrix([[1, 2], [3, 1]])
    # 1   2
    # 3   1
    # B=  0   -2
    #     -3  0
    assert not r["row_dominant"]  # 1<2 đường chéo nhỏ hơn -> không trội hàng
    assert not r["col_dominant"]  # 1<3 đường chéo nhỏ hơn -> không trội cột
    assert gan_bang(r["q"], 3.0)  # max(2/1 , 3/1)=3
    assert r["Bankinhpho"] > 1
    assert gan_bang(r["Bankinhpho"], 2.449490)  # căn 6 > 1
    # hI= h   0
    #     0   h
    # B-hI=     -h  -2
    #           -3   -h
    # det(B-hI)= h^2-6=0  -> h=±căn6
    assert r["q_gaussseidel"] == float("inf")
    # nếu mẫu âm thì không dùng được đánh giá q_gs, gán = inf (vô cùng) để không đảm bảo hội tụ. Vì cần q_gs < 1
    # nếu mẫu bé hơn 0 thì không chia được / không đánh giá co được
    # vì sao phải chặn? Ma trận A không hề trội, bán kính phổ lớn hơn 1. Nếu cứ chia máy sẽ trả kết luận giả
    # q_gs = vô cùng thì phân kì? Không kết luận được, cần xét thêm bán kính phổ nhỏ hơn 1
    # còn nếu A chéo trội nghiêm ngặt thì chắc chắn mẫu dương


# 5 chéo trội không nghiêm ngặt (lớn hơn hoặc bằng)
def test_cheo_troi_dau_bang():
    r = analyze_matrix([[2, 2], [1, 3]])
    # A = 2   2
    #     1   3
    assert r["row_dominant"]
    assert not r["row_strict"]  # hàng 1: 2 = 2


# 6 Đường chéo âm (dùng trị tuyệt đối khi so sánh)
def test_duong_cheo_am():
    r = analyze_matrix([[-4, 1], [1, -4]])
    # A = -4   1
    #      1  -4
    # B = 0      0,25
    #     0,25     0
    # hI = h     0
    #      0     h
    # B-hI = -h   0,25
    #        0,25  -h
    # det(B-hI)= h^2 - 0,0625 -> h= ±0,25
    assert r["row_strict"]  # |-4|>1 chéo trội nghiêm ngặt
    assert gan_bang(r["q"], 0.25)
    assert gan_bang(r["Bankinhpho"], 0.25)
    assert gan_bang(r["matranB"][0][1], 0.25)
    # dấu của B phải đúng: -1 / -4 = +0.25  (hàng 0 cột 1 phải là 0,25)


# 7 Giá trị riêng phức
def test_gia_tri_rieng_phuc():
    r = analyze_matrix([[1, 2], [-2, 1]])
    # A = 1    2
    #    -2    1
    # B=  0    -2
    #     2    0
    # hI= h   0
    #     0   h
    # B-hI= -h   -2
    #        2   -h
    # det(B-hI)= h^2 + 4 -> h=2i và -2i
    # but |2i|=2. Test để chắc chắn máy dùng np.abs đúng cho số phức
    assert gan_bang(r["Bankinhpho"], 2.0)


# 8 Kiểm tra làm tròn 6 chữ số
def test_lam_tron_6_chu_so():
    r = analyze_matrix([[4, 1, 1], [1, 5, 2], [0, 1, 3]])
    pho = r["Bankinhpho"]
    assert gan_bang(pho, lam_tron_6(pho))


# 9 bkpho(B) <= q, kiểm tra 1205 trường hợp khác nhau để bkpho luôn nhỏ hơn hoặc bằng q
def test_bkpho_nho_hon_hoac_bang_q():
    # rng cơ chế giống random, sinh ra ma trận A ngẫu nhiên
    # Dùng công thức sinh giả ngẫu nhiên tự cài đặt, không gọi hàm random của thư viện
    seed = 0
    for _ in range(1205):
        A = []
        for i in range(4):
            row = []
            for j in range(4):
                seed = (1103515245 * seed + 12345) % 2147483648
                value = 1 + (seed / 2147483648) * 4
                if i == j:
                    value += 10
                row.append(value)
            A.append(row)
        # seed 0 là để cố định kết quả ngẫu nhiên
        # sau 1205 lần sinh ra 1205 ma trận khác nhau, nhưng chạy lại vẫn ra đúng 1205 ma trận đó
        r = analyze_matrix(A)
        assert r["Bankinhpho"] <= r["q"] + 1e-6


# 10 Đường chéo của B luôn bằng 0, chuanB = q
def test_cheo_B_bang_0_va_chuanB_bang_q():
    A = [[10, 2, 1], [3, 8, 2], [1, 1, 5]]
    # tính B như trên
    r = analyze_matrix(A)
    assert gan_bang(r["matranB"][0][0], 0)
    assert gan_bang(r["matranB"][1][1], 0)
    assert gan_bang(r["matranB"][2][2], 0)
    assert gan_bang(r["chuanB"], r["q"])


# 11 Đầu vào nguyên (int) vẫn chạy đúng
# hạn chế trường hợp phép chia lấy phần nguyên
# vd: 1/3 mà bị làm tròn thành 0 thay vì 0,333333333333 thì hệ số q sẽ sai
def test_dau_vao_so_nguyen():
    r = analyze_matrix([[3, 1], [1, 3]])
    assert gan_bang(r["q"], 1/3)
    # kiểm tra đã đổi dữ liệu đầu vào sang số thực dtype=float chưa


# 12 Ma trận 1x1
def test_ma_tran_1x1():
    r = analyze_matrix([[12052006]])
    assert r["q"] == 0.0
    assert r["Bankinhpho"] == 0.0
    assert r["doixungg"]


# 13-16 Các trường hợp phải báo lỗi
def test_khong_vuong():
    try:
        analyze_matrix([[1, 2, 3], [4, 5, 6]])
    except ValueError:
        return
    assert False, "Phải phát sinh ValueError"


def test_khong_phai_ma_tran_2_chieu():
    try:
        analyze_matrix([1, 2, 3])
    except (ValueError, TypeError):
        return
    assert False, "Phải phát sinh lỗi"


def test_duong_cheo_bang_0():
    try:
        analyze_matrix([[0, 1], [1, 2]])
    except ValueError:
        return
    assert False, "Phải phát sinh ValueError"


def test_duong_cheo_rat_nho():
    try:
        analyze_matrix([[1e-15, 1], [1, 2]])
    except ValueError:
        return
    assert False, "Phải phát sinh ValueError"

### CA BIÊN TRONG SÁCH THEO YÊU CẦU CỦA CÔ ####

def test_khong_cheo_troi_nhung_hoi_tu():
# Xét ma trận A = [[2.0, 1.9], 
#                  [1.9, 2.0]]
# - Tổng phần tử ngoài đường chéo hàng 1: |1.9| = 1.9 gần bằng |2.0| nhưng 2.0 > 1.9 
# - Xét ma trận B = [[0.0, -0.95], 
#                    [-0.95, 0.0]]
# - Giá trị riêng của B: 0.95 => Bán kính phổ = 0.95 < 1.
# -> Kết luận: Không chéo trội nhưng phương pháp lặp vẫn hội tụ.
    A = [[2.0, 1.9],
         [1.9, 2.0]]
    r = analyze_matrix(A)
    assert r["row_strict"]
    assert r["Bankinhpho"] < 1.0
    assert gan_bang(r["Bankinhpho"], 0.95)
    assert gan_bang(r["q"], 0.95)
def test_ma_tran_lap_phan_ky():
#xét ma trận A = [[1.0, 2.0], 
#                  [2.0, 1.0]]
# - Phần tử đường chéo |1.0| < tổng ngoài đường chéo |2.0| -> Không chéo trội.
# - Ma trận B = [[0.0, -2.0], 
#                [-2.0, 0.0]]
# - Giá trị riêng của B:  2.0 => Bán kính phổ = 2.0 >= 1.
# -> Kết luận: Bán kính phổ  >= 1 làm dãy lặp phân kỳ.
    A = [[1.0, 2.0],
        [2.0, 1.0]]
    r = analyze_matrix(A)
    assert not r["row_dominant"]
    assert not r["col_dominant"]
    assert r["Bankinhpho"] >= 1.0
    assert gan_bang(r["Bankinhpho"], 2.0)

test_functions = [
    test_vi_du_chuan, test_ma_tran_duong_cheo, test_doi_xung_2x2,
    test_khong_cheo_troi, test_cheo_troi_dau_bang, test_duong_cheo_am,
    test_gia_tri_rieng_phuc, test_lam_tron_6_chu_so, test_bkpho_nho_hon_hoac_bang_q,
    test_cheo_B_bang_0_va_chuanB_bang_q, test_dau_vao_so_nguyen,
    test_ma_tran_1x1, test_khong_vuong, test_khong_phai_ma_tran_2_chieu,
    test_duong_cheo_bang_0, test_duong_cheo_rat_nho,
    test_khong_cheo_troi_nhung_hoi_tu, test_ma_tran_lap_phan_ky
]

for func in test_functions:
    func()
    print(f"✔ {func.__name__} ---> PASSED")

print("pass")
```
